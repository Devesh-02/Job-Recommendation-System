```python
import pandas as pd
import os
import json

# Path to the dataset folder
dataset_path = r"C:\Users\Dev\.cache\kagglehub\datasets\promptcloud\job-dataset-indeed-india\versions\1"
file_name = "marketing_sample_for_indeed_co_in-indeed_co_in_job_data__20210701_20210930__30k_data.ldjson"
file_path = os.path.join(dataset_path, file_name)

# Read LDJSON file line by line safely
records = []
with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        try:
            records.append(json.loads(line))
        except Exception as e:
            # Skip bad lines
            print("Skipping line due to error:", e)

# Convert to DataFrame
df = pd.DataFrame(records)

# Optional: check columns
print("Columns:", df.columns)

# Save as CSV for easier use
output_csv_path = os.path.join(dataset_path, "indeed_india_jobs_clean.csv")
df.to_csv(output_csv_path, index=False, encoding="utf-8")

print(f"✅ Dataset saved as CSV at: {output_csv_path}")
print("Shape:", df.shape)
```

```python
import pandas as pd
import numpy as np

# Load dataset
file_path = r"C:\Users\Dev\.cache\kagglehub\datasets\promptcloud\job-dataset-indeed-india\versions\1\indeed_india_jobs_clean.csv"
df = pd.read_csv(file_path)

print("Initial shape:", df.shape)

# 1. Remove duplicates
df.drop_duplicates(inplace=True)
print("After removing duplicates:", df.shape)

#  2. Convert list-like objects into strings
for col in df.columns:
    if df[col].apply(lambda x: isinstance(x, list)).any():  
        df[col] = df[col].astype(str)

# 3. Handle missing values
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:  # numeric columns
        df[col].fillna(df[col].mean(), inplace=True)
    else:  # categorical columns
        try:
            df[col].fillna(df[col].mode()[0], inplace=True)
        except:
            df[col].fillna("Unknown", inplace=True)

print("Missing values handled.")

# 4. Handle outliers using IQR (only for numeric columns)
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = np.where(df[col] < lower, lower,
                       np.where(df[col] > upper, upper, df[col]))

print("Outliers handled.")

# Save the cleaned dataset
cleaned_file = r"C:\Users\Dev\.cache\kagglehub\datasets\promptcloud\job-dataset-indeed-india\versions\1\indeed_india_jobs_cleaned_final.csv"
df.to_csv(cleaned_file, index=False)

print(f"✅ Final cleaned dataset saved at: {cleaned_file}")
print("Final shape:", df.shape)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# --------------------------
# Load cleaned dataset
# --------------------------
file_path = r"C:\Users\Dev\.cache\kagglehub\datasets\promptcloud\job-dataset-indeed-india\versions\1\indeed_india_jobs_clean.csv"
df = pd.read_csv(file_path)

# --------------------------
# Text preprocessing (if not done)
# --------------------------
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    if pd.isnull(text):
        return ""
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = text.split()
    tokens = [w for w in tokens if w not in stop_words]
    tokens = [lemmatizer.lemmatize(w) for w in tokens]
    return " ".join(tokens)

df["clean_job_title"] = df["job_title"].astype(str).apply(clean_text)
df["clean_job_description"] = df["job_description"].astype(str).apply(clean_text)

# Combine title + description
df["combined_text"] = df["clean_job_title"] + " " + df["clean_job_description"]

# --------------------------
# TF-IDF Vectorization
# --------------------------
vectorizer = TfidfVectorizer(max_features=5000)
tfidf_matrix = vectorizer.fit_transform(df["combined_text"])

# --------------------------
# Nearest Neighbors for Recommendation
# --------------------------
nn_model = NearestNeighbors(metric='cosine', algorithm='brute')
nn_model.fit(tfidf_matrix)

def recommend_jobs(job_index, num_recommendations=5):
    distances, indices = nn_model.kneighbors(tfidf_matrix[job_index], n_neighbors=num_recommendations+1)
    recommended_indices = indices[0][1:]  # skip itself
    return df.iloc[recommended_indices][["job_title", "company_name", "city", "state", "category"]]

# Example: recommend for job index 10
print("Query Job:\n", df.iloc[10][["job_title", "company_name", "city"]])
print("\nRecommended Jobs:\n", recommend_jobs(10, num_recommendations=5))
```

```python
from difflib import get_close_matches

def recommend_by_title(job_title_input, num_recommendations=5):
    # Find closest match to user input
    job_title_input = job_title_input.lower()
    all_titles = df["clean_job_title"].tolist()
    closest_match = get_close_matches(job_title_input, all_titles, n=1, cutoff=0.5)
    
    if not closest_match:
        return "No matching job title found!"
    
    # Get index of the matched job
    job_index = df[df["clean_job_title"] == closest_match[0]].index[0]
    
    # Recommend jobs
    return recommend_jobs(job_index, num_recommendations)

# Example: user types a job title
user_input = "data scientist"
print(f"Recommendations for '{user_input}':\n")
print(recommend_by_title(user_input, num_recommendations=5))
```
